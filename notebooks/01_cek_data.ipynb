```python
import pandas as pd

pd.set_option("display.max_columns", None)  # tampilkan semua kolom, jangan disembunyikan "..."

df = pd.read_parquet("../data/raw/hpd_2025_01.parquet")
df.head(10)
```

```python
df["durasi_hari"] = (df["closed_date"] - df["created_date"]).dt.total_seconds() / 86400

ringkas = (
    df.dropna(subset=["durasi_hari"])
      .groupby("complaint_type")["durasi_hari"]
      .describe(percentiles=[0.5, 0.75, 0.9])
      .sort_values("count", ascending=False)
      .round(1)
)
ringkas.head(10)
```
